# Notebook 02: Analysis of Indexed Legal Corpus

This notebook analyzes entity/relation distribution, graph statistics, and degree distribution for the full LightRAG indexed corpus.

```python
import os
import sys
import json
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

sys.path.insert(0, os.path.abspath("../src"))
from utils import load_config
```

## 1. Inspect Working Directory KV Stores

```python
config = load_config("../config/lightrag_config.example.env")
working_dir = os.path.abspath(config.working_dir)

text_chunks_file = os.path.join(working_dir, "kv_store_text_chunks.json")
full_docs_file = os.path.join(working_dir, "kv_store_full_docs.json")

if os.path.exists(text_chunks_file):
    with open(text_chunks_file, "r", encoding="utf-8") as f:
        chunks = json.load(f)
    print(f"Total Text Chunks in Index: {len(chunks)}")
else:
    print("Text chunks file not found.")
```

## 2. Graph Node & Edge Degree Metrics

```python
graphml_file = os.path.join(working_dir, "graph_chunk_entity_relation.graphml")
if os.path.exists(graphml_file):
    G = nx.read_graphml(graphml_file)
    degrees = [d for _, d in G.degree()]
    df_deg = pd.DataFrame(degrees, columns=["degree"])
    print(df_deg.describe())
    
    plt.figure(figsize=(8, 4))
    plt.hist(degrees, bins=20, color='teal', edgecolor='black')
    plt.title("Entity Degree Distribution")
    plt.xlabel("Degree (Connections)")
    plt.ylabel("Count")
    plt.show()
```

# Notebook 03: Query Performance & Quality Evaluation

This notebook visualizes and analyzes evaluation metrics (ROUGE-L, Recall, Latency) across different retrieval baselines and LightRAG query modes.

```python
import os
import json
import pandas as pd
import matplotlib.pyplot as plt

eval_results_file = "../output/evaluation_results.json"

if os.path.exists(eval_results_file):
    with open(eval_results_file, "r", encoding="utf-8") as f:
        data = json.load(f)
    
    summary = data.get("summary", {})
    df_summary = pd.DataFrame.from_dict(summary, orient="index")
    print("=== Evaluation Summary Table ===")
    display(df_summary)
else:
    print("Evaluation results file not found. Run src/evaluate_lightrag.py first.")
```

## 1. ROUGE-L Comparison Chart

```python
if os.path.exists(eval_results_file) and not df_summary.empty:
    fig, ax = plt.subplots(1, 2, figsize=(14, 5))
    
    # Chart 1: ROUGE-L
    df_summary["avg_rouge_l"].plot(kind="bar", ax=ax[0], color="royalblue", edgecolor="black")
    ax[0].set_title("Average ROUGE-L Score by Baseline")
    ax[0].set_ylabel("ROUGE-L")
    ax[0].set_ylim(0, 1.0)
    ax[0].grid(axis="y", linestyle="--", alpha=0.7)
    
    # Chart 2: Latency
    df_summary["avg_latency_ms"].plot(kind="bar", ax=ax[1], color="crimson", edgecolor="black")
    ax[1].set_title("Average Query Latency (ms)")
    ax[1].set_ylabel("Latency (ms)")
    ax[1].grid(axis="y", linestyle="--", alpha=0.7)
    
    plt.tight_layout()
    plt.show()
```
